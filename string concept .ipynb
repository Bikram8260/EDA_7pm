# String Method

| Method              | Description                              | Example                                   |
| ------------------- | ---------------------------------------- | ----------------------------------------- |
| `str.lower()`       | Converts to lowercase                    | `'HELLO'.lower()` → `'hello'`             |
| `str.upper()`       | Converts to uppercase                    | `'hello'.upper()` → `'HELLO'`             |
| `str.title()`       | Capitalizes first letter of each word    | `'hello world'.title()` → `'Hello World'` |
| `str.strip()`       | Removes leading/trailing spaces          | `'  hi  '.strip()` → `'hi'`               |
| `str.replace(a, b)` | Replaces `a` with `b`                    | `'cat'.replace('c', 'b')` → `'bat'`       |
| `str.split()`       | Splits string into list by spaces        | `'a b c'.split()` → `['a', 'b', 'c']`     |
| `str.join(list)`    | Joins list elements into a string        | `' '.join(['a', 'b'])` → `'a b'`          |
| `str.count('sub')`  | Counts occurrences of substring          | `'ola ola'.count('ola')` → `2`            |
| `str.find('sub')`   | Finds index of first occurrence          | `'apple'.find('p')` → `1`                 |
| `str.startswith(x)` | Returns `True` if string starts with `x` | `'python'.startswith('py')` → `True`      |
| `str.endswith(x)`   | Returns `True` if string ends with `x`   | `'hello'.endswith('lo')` → `True`         |
| `str.isalpha()`     | Checks if all chars are letters          | `'abc'.isalpha()` → `True`                |
| `str.isdigit()`     | Checks if all chars are digits           | `'123'.isdigit()` → `True`                |

# List Method

| Method               | Description                              | Example                            |
| -------------------- | ---------------------------------------- | ---------------------------------- |
| `append(x)`          | Adds item `x` at the end                 | `fruits.append('apple')`           |
| `insert(i, x)`       | Inserts item `x` at index `i`            | `fruits.insert(1, 'banana')`       |
| `extend(list2)`      | Adds all items from another list         | `fruits.extend(['mango', 'kiwi'])` |
| `remove(x)`          | Removes first occurrence of `x`          | `fruits.remove('apple')`           |
| `pop()`              | Removes and returns last item            | `fruits.pop()`                     |
| `pop(i)`             | Removes and returns item at index `i`    | `fruits.pop(1)`                    |
| `clear()`            | Removes all items from the list          | `fruits.clear()`                   |
| `index(x)`           | Returns index of first occurrence of `x` | `fruits.index('banana')`           |
| `count(x)`           | Counts how many times `x` occurs         | `fruits.count('apple')`            |
| `sort()`             | Sorts the list in ascending order        | `numbers.sort()`                   |
| `sort(reverse=True)` | Sorts in descending order                | `numbers.sort(reverse=True)`       |
| `reverse()`          | Reverses the list order                  | `fruits.reverse()`                 |
| `copy()`             | Returns a copy of the list               | `new_list = fruits.copy()`         |
