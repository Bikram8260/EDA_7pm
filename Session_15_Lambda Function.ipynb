##### Case-1: function with one argument

```python
def add(a):
    return(a+10)
add(20)
```

```python
# what is the function: add()
# what is the argument: a
# what is the output: a+10
<function_name>= lambda <arg> :<output>
```

```python
add= lambda a : a+10
add(20)
```

#### case-2: multiple argument

```python
def avg(x,y,z):
    return((x+y+z)/3)
avg(10,20,30)
```

```python
avg= lambda x,y,z : (x+y+z)/3
avg(10,20,30)
```

```python
avg= lambda x,y,z : round((x+y+z)/3,2)
avg(10,20,35)
```

```python
avg= lambda x,y,z=30 : (x+y+z)/3
avg(10,20)
```

#### Case-3: if-else

```python
def even_odd(num):
    if num%2==0:
        return('even')
    else:
        return('odd')
even_odd(10)
```

```python
even_odd= lambda num : 'even' if num%2==0 else 'odd'
even_odd(20)
```

```python
# step-3: map the function and list
# map(< function>,<list>)
map(lambda i:i.title(),Input)
```

```python
# step-4: list
# list(map(< function>,<list>))
list(map(lambda i:i.title(),Input))
```

```python
Input=['hyd','bengalur','mumbai','chenai']
list(map(lambda i: i.upper(),Input))
```

```python
Input=['hyd','bengalur','mum#bai','chen#ai']
list(map(lambda i: '#' in i ,Input))
```

### Filter

```python
Input=['hyd','bengalur','mum#bai','chen#ai']
list(filter(lambda i: '#' in i ,Input))
```
