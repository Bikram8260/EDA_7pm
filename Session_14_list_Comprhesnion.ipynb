$Case-3$:** for-if-else

```python
[<output> <for loop>]
[<output> <for loop> <if condition>]
[<output if> <if cond> else <else output> <for loop>]
```

```python
import random
numbers=[random.randint(1,100) for i in range(5)]
output=[]
for i in numbers:
    if i%2==0:
        output.append(f'even:{i}')
    else:
        output.append(f'odd:{i}')
        
output
```

```python
import random
numbers=[random.randint(1,100) for i in range(5)]
output1=[f'even:{i}' if i%2==0 else f'odd: {i}' for i in numbers]
output1
```

$Case-4$:if-elif-else

```python
[<output> <for loop>]
[<output> <for loop> <if condition>]
[<output if> <if cond> else <else output> <for loop>]
[<output if> <if cond> \
 else <elif output> <if elifcond> \
 else <else output> <for loop>]
```

```python
l=[50,70,90]
out=[]
for i in l:
    if i>85:
        out.append('A')
    elif i>65:
        out.append('B')
    else:
        out.append('c')
        
out
```

```python
l=[50,70,90]
out=[f'A' if i>85 \
     else f'B' if i>65 \
     else f'c' for i in l]
out
```
